```python

import pandas as pd

from hf_dataset_loader import HFDatasetLoader
from dataset_summarizer import DatasetSummarizer


  
dataset_id = "scikit-learn/iris"

loader = HFDatasetLoader(dataset_id, split="train")

df = loader.load_data()

print("Dataset cargado correctamente.")
print("Dimensiones:", df.shape)

display(df.head())


df = df.rename(
    columns={"sepal_length": "longitud_sepalo"}
)

print("Columnas después de la transformación:")
print(df.columns.tolist())

display(df.head())


summarizer = DatasetSummarizer(df)

summarizer.generate_summary("resumen_dataset.txt")
  

with open("resumen_dataset.txt", "r", encoding="utf-8") as file:
    contenido = file.read()

print(contenido)
  
```
